Target Distribution

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=cleaned_df,
    x='category',
    order=cleaned_df['category'].value_counts().index,
    palette='Set2'
)

plt.xticks(rotation=45)

plt.title("Incident Category Distribution")

plt.show()
```

Priority Distribution

```python
plt.style.use('tableau-colorblind10')
plt.figure(figsize=(10,6))
sns.countplot(data=cleaned_df,x='priority')
plt.title("Incident Priority Distribution")
```

```python
plt.figure(figsize=(6,4))
plt.style.use('seaborn-v0_8-pastel')
sns.countplot(
    data=cleaned_df,
    x='is_resolved'
)

plt.title("Resolved vs Unresolved")

plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.style.use('tableau-colorblind10')
sns.countplot(
    data=cleaned_df,
    x='impact',
)

plt.title("Impact Distribution")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=cleaned_df,
    x='urgency',
    palette='Oranges'
)

plt.title("Urgency Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=cleaned_df,
    y='assignment_group',
    order=cleaned_df['assignment_group'].value_counts().head(15).index,
    palette='crest'
)

plt.title("Top Assignment Groups")

plt.show()
```

```python
plt.figure(figsize=(8,6))

corr = cleaned_df.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```
