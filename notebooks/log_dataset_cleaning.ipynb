Changing the date time values format to datetime

```python
date_columns=['opened_at','sys_created_at','sys_updated_at','resolved_at']
for col in date_columns:
    df[col]=pd.to_datetime(df[col],errors='coerce')
```

```python
df.drop(columns=['cmdb_ci'],axis=1,inplace=True,errors='ignore')
```

```python
#creating a temporary missing dataframe
missing=pd.DataFrame({'missing':df.isnull().sum(),
                      'missing_percentage':df.isnull().mean()*100})
missing=missing.sort_values(by='missing_percentage',ascending=False)
print(missing)
```
